## 1) `while` loop syntax and indentation

A `while` loop has the following general form. This is a syntax pattern; the runnable examples below use actual conditions and instructions.

```python
while condition:
    # Instructions to repeat
```

Read this as:

> **While the condition is True, repeat the indented instructions.**

- The colon `:` is required.
- The **indented block** is the loop body.
- Python checks the condition **before every repetition**.
- When the condition becomes `False`, Python skips the loop body and continues with the code after the loop.

One repetition of the loop body is called an **iteration**.

### Indentation tells Python what belongs inside the loop

Use **four spaces** for statements directly inside the loop.

- Statements indented under the `while` line belong to its loop body and repeat whenever the condition is `True`.
- Returning to the original indentation level marks the end of the loop body.
- A statement after the loop, at the original indentation level, runs after the loop finishes.
- A blank line helps readability, but **indentation—not the blank line—determines where the loop body ends**.


## 2) A first simple example

Print the numbers 1, 2, and 3, followed by `Done`.

Here, `number <= 3` means: **Is `number` less than or equal to 3?**

Notice which statements are indented and which are not.


````python
number = 1

while number <= 3:
    print(number)          # Inside the loop: repeats
    number = number + 1    # Inside the loop: repeats

print("Done")              # Outside the loop: runs after the loop finishes
````

**Expected output:**

```text
1
2
3
Done
```

### How does Python execute this code?

| Condition check | Current `number` | Is `number <= 3` True? | What happens? |
| --- | ---: | --- | --- |
| First | 1 | True | Print 1, then change `number` to 2 |
| Second | 2 | True | Print 2, then change `number` to 3 |
| Third | 3 | True | Print 3, then change `number` to 4 |
| Fourth | 4 | False | Exit the loop |

Both `print(number)` and `number = number + 1` are indented, so both repeat.

The statement `print("Done")` is **not indented**. It is outside the loop and prints **once, after the loop finishes**.


### Understanding `number = number + 1`

This is an **assignment statement**, not a mathematical equation.

Python:

1. Reads the current value of `number` on the right side.
2. Adds 1.
3. Stores the result back in `number` on the left side.

For example, if `number` is currently 2, this statement changes it to 3.

> In this lecture, we will write updates explicitly, such as `number = number + 1`, so that each step is easy to follow.


### Can a `while` loop run zero times?

Yes. In the next example, `number` starts at 5.

The first condition check, `5 <= 3`, is `False`. Python skips the loop body and prints only `Done`.


````python
number = 5

while number <= 3:
    print(number)
    number = number + 1

print("Done")
````

## 3) Starting value, condition, and update

Many beginner `while` loops have three important parts:

| Part | In our first example | Purpose |
| --- | --- | --- |
| Starting value | `number = 1` | Establish where the loop starts |
| Condition | `number <= 3` | Decide whether to repeat |
| Update | `number = number + 1` | Change the value used in the condition |

### What happens if we forget the update?

If we remove `number = number + 1` from the first example, `number` stays at 1.

The condition `number <= 3` remains `True`, so the program keeps printing 1.

This is an **infinite loop**: a loop that does not stop on its own.

> Always ask: **What changes inside this loop that will eventually make the condition False?**

If you accidentally create an infinite loop in a notebook, interrupt the running cell.


## 4) A finance example: saving toward a goal

Suppose:

- You start with **\$0**.
- You deposit **\$200 each month**.
- Your savings goal is **\$1,000**.
- For this example, we ignore interest.

How many months will it take to reach your goal?


````python
balance = 0
month = 0

while balance < 1000:
    balance = balance + 200
    month = month + 1

    print(f"Month {month}: ${balance}")

print(f"Months needed: {month}")
````

**Expected output:**

```text
Month 1: $200
Month 2: $400
Month 3: $600
Month 4: $800
Month 5: $1000
Months needed: 5
```

**Explanation:**

- `balance = balance + 200` adds this month's deposit to the current balance.
- `month = month + 1` counts one more month.
- `while balance < 1000:` keeps repeating while the balance is below the goal.

After the fifth deposit, the balance is 1,000. Python checks `1000 < 1000`, which is `False`, so the loop stops.

**Indentation check:** The monthly balance prints inside the loop. The final number of months prints outside the loop, once.

**Predict before running:** If the monthly deposit changes from \$200 to \$250, how many months will be needed?

**Answer:** 4 months. Change `200` to `250` in the code cell above and run it again to check.

## 6) Practice 2: calculate a running total

A **running total** is a total that grows as we add each new value.

The statement `total = total + number` means:

> Add the current value of `number` to the existing total, and store the result back in `total`.

**Task:** Use a `while` loop to calculate:

```text
1 + 2 + 3 + 4 + 5
```

**Expected output after completing the exercise:**

```text
Total: 15
```

Complete the starter cell below. Until you add your loop, it prints `Total: 0`.


````python
# Practice 2: calculate a running total
number = 1
total = 0

# Write your while loop below.


print(f"Total: {total}")
````

### Practice 2 solution — after you try it


````python
number = 1
total = 0

while number <= 5:
    total = total + number
    number = number + 1

print(f"Total: {total}")
````

### Trace the running total

| Iteration | Number being added | Total before addition | Total after addition |
| --- | ---: | ---: | ---: |
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |
| 5 | 5 | 10 | 15 |

**Why is `total = 0` before the loop?**

We initialize the total **once**, before adding any numbers.

If we put `total = 0` inside the loop, we would erase the accumulated total during every repetition.

> Use `total` as the variable name. Avoid naming a variable `sum`, because Python already has a built-in function named `sum()`.


## 7) Common mistakes

| Mistake | What to remember |
| --- | --- |
| Forgetting the colon | Write `while condition:` |
| Incorrect indentation | Use four spaces for statements directly inside the loop |
| Forgetting to update the condition's variable | The loop may never stop |
| Resetting a running total inside the loop | Initialize it before the loop |
| Printing a final result inside the loop | Put it after the loop if it should appear only once |
| Confusing `<` with `<=` | Decide whether the boundary value should be included |


## 8) Final example: enter midterm scores and calculate the average

Suppose we want to calculate the average midterm score for a class.

We do **not** know in advance how many scores the user will enter.

Our program should:

1. Ask the user to enter a midterm score.
2. Add the score to a running total.
3. Count the score.
4. Ask for another score.
5. Stop when the user enters `q` or `Q`.
6. Calculate the average.

This is a useful application of a `while` loop: **the user decides how many times the task repeats**.

**Assumption for this example:** Enter at least **one valid numeric score** before entering `q` or `Q`. Each score should be between 0 and 100.


### 8.1 Two variables we need

We start with `total = 0` and `count = 0`.

- `total` stores the sum of all scores entered so far.
- `count` stores how many scores have been entered.

Each time we receive a score, we use these two statements:

- `total = total + score`: add the new score to the existing total.
- `count = count + 1`: count one more score.

For example, after entering 80 and 90, `total` is 170.0 and `count` is 2.

The average is calculated with `average = total / count`.


### 8.2 Remember: `input()` returns text

We use `input()` to display the prompt and store the user's response in `score_text`.

- If the user types `85`, Python stores the text `"85"`.
- If the user types `q`, Python stores the text `"q"`.

Before doing arithmetic, we convert numeric text to a number with `score = float(score_text)`.

We use `float()` so that scores such as `85.5` are also allowed.

**We must check for `q` or `Q` before converting.** Python cannot convert `"q"` to a numeric score.


### 8.3 The loop condition and sentinel values

The loop condition is:

```python
while score_text != "q" and score_text != "Q":
```

- `!=` means **not equal to**.
- `and` means **both comparisons must be True**.

Read the condition as:

> Keep repeating while the input is neither lowercase `q` nor uppercase `Q`.

| Input | Not equal to `"q"`? | Not equal to `"Q"`? | Repeat the loop? |
| --- | --- | --- | --- |
| `"85"` | True | True | Yes |
| `"q"` | False | True | No |
| `"Q"` | True | False | No |

A **sentinel value** is a special input that tells the loop to stop. In this example, `"q"` and `"Q"` are sentinel values. They signal that the user has finished entering scores and are **not included in the total or count**.


### 8.4 Complete program

Run the next cell and enter `80`, `90`, `100`, and then `q`, one entry at a time.

You can run the cell again with other scores. Each run resets `total` and `count` to 0.

**Remember:** Enter at least one valid score before quitting.


````python
# Initialize the running total and count.
total = 0
count = 0

# Get the first input.
score_text = input("Enter the midterm score (enter q or Q to quit): ")

while score_text != "q" and score_text != "Q":
    # Convert the input to a number.
    score = float(score_text)

    # Add the score to the total and count one more score.
    total = total + score
    count = count + 1

    # Ask for the next score.
    score_text = input("Enter the midterm score (enter q or Q to quit): ")

# Outside the loop: calculate and display the results.
average = total / count

print(f"Total: {total}")
print(f"Number of scores: {count}")
print(f"Average: {average}")
````

### 8.5 Example run

```text

Enter the midterm score (enter q or Q to quit): 100
Enter the midterm score (enter q or Q to quit): q

Number of scores: 3
Average: 90.0
```

### Trace the program

| Input | What happens? | Running total | Count |
| --- | --- | ---: | ---: |
| `80` | Add 80 and count one score | 80.0 | 1 |
| `90` | Add 90 and count one score | 170.0 | 2 |
| `100` | Add 100 and count one score | 270.0 | 3 |
| `q` | Exit the loop | 270.0 | 3 |

The quit input is **not** added to the total and is **not** counted as a score.


### 8.6 Why are there two `input()` statements?

The first input appears **before the loop**. It gives Python something to check when it reaches the first `while` condition.

The second input appears **inside the loop**, after processing the current score. It asks for the next score or the quit command.

If we forget the second input, the program will keep processing the same score repeatedly.


### 8.7 Indentation in the final example

These statements are indented inside the loop, so they repeat:

- Convert the numeric text to a score.
- Add the score to `total`.
- Increase `count` by 1.
- Ask for the next input.

The statement `average = total / count` and the final three `print()` statements are **not indented**. They run **once, after the user enters `q` or `Q` and the loop ends**.
