## 1. What is a class?

You have already been using objects: `"aapl".upper()` and `prices.append(107.3)` call *methods* on a string object and a list object. The dot is the giveaway. Today we learn to design our own kinds of objects.

> **A class is a blueprint (a template).** It describes a *kind* of thing: what data it holds and what it can do.

| Part | What it is | `Student` example |
|---|---|---|
| **Attributes** | The data / characteristics (variables that belong to each object) | `name`, `major`, `gpa` |
| **Methods** | The actions / behaviors (functions that belong to the class) | `study()`, `play()` |

**Analogy:** a blank loan application form is the *class*; each customer's completed application is an *object*. The form describes what every application contains, but the form itself is not anyone's application.

Here is a `Student` class with three attributes and two methods. Read it top to bottom; we dissect it right after.

````python
class Student:

    # __init__ is a special method that sets up the ATTRIBUTES (the data) of each new student
    def __init__(self, name, major, gpa):
        self.name = name        # attribute 1
        self.major = major      # attribute 2
        self.gpa = gpa          # attribute 3

    # A METHOD: something a student can DO
    def study(self):
        print(f"{self.name} is studying {self.major}.")

    # Another METHOD
    def play(self):
        print(f"{self.name} is taking a break and playing.")
````

### Anatomy of the class

| Line | Meaning |
|---|---|
| `class Student:` | Start defining a class called `Student`. Everything indented below belongs to it. |
| `def __init__(self, name, major, gpa):` | The **initializer** (also called the *constructor*). Python calls it automatically every time a new student is created. Its job: store the starting data. |
| `self.name = name` | Create an **attribute** called `name` on the object and store the value that was passed in. |
| `def study(self):` | A **method**: a function defined inside the class. Its first parameter is always `self`. |
| `self` | Means "**this particular object**" — the specific student we are working with right now. More on this soon. |

> **Naming convention:** class names use `CapitalizedWords` (`Student`, `BankAccount`); variables, functions, and methods use `snake_case` (`gpa`, `update_price`).

**Important: defining a class creates no students.** Running the cell above printed nothing: we only drew the blueprint. To get an actual student, we need to create an **object**.

## 2. What is an object?

> **An object is a specific, concrete thing created from a class.** It is also called an **instance** of the class, and creating one is called **instantiation**.

If `Student` is the blueprint, then *Alice, a Finance major with a 3.8 GPA* is an object.

### How to create an object from a class

Call the class name **as if it were a function**, passing the values `__init__` asks for (everything except `self`):

````python
s1 = Student("Alice", "Finance", 3.8)     # create a Student object and store it in s1
````

### What happens when that line runs? (When is `__init__` called?)

```text
s1 = Student("Alice", "Finance", 3.8)
        |
        |-- Step 1. Python creates a brand-new, empty Student object.
        |
        |-- Step 2. Python AUTOMATICALLY calls __init__ on that object:
        |              Student.__init__(new_object, "Alice", "Finance", 3.8)
        |                                   ^self     ^name    ^major   ^gpa
        |          Inside __init__:
        |              self.name  = "Alice"      -> the object now has a .name
        |              self.major = "Finance"    -> the object now has a .major
        |              self.gpa   = 3.8          -> the object now has a .gpa
        |
        '-- Step 3. The finished object is stored in the variable s1.
```

**Key facts about `__init__`:**

1. It is called **automatically**. You never write `s1.__init__(...)` yourself.
2. It is called **exactly once per object**, at the moment the object is created — i.e., when the line `Student(...)` runs.
3. It is **not** called when you *define* the class. Defining the class only draws the blueprint.
4. The name is "init" with **two underscores on each side**: `__init__` (short for *initialize*). Double underscores mark special methods that Python calls on its own.

Let's *see* facts 2 and 3 with a tiny class whose `__init__` announces itself:

````python
class Demo:
    def __init__(self):
        print("   >>> __init__ was just called!")

print("Demo class defined. Nothing printed above this line -> __init__ did NOT run.")
d1 = Demo()
d2 = Demo()
print("Two objects created -> __init__ ran twice, once per object.")
````

### What is `self`?

`self` is how a method refers to **"the object I am currently working on."**

- When `__init__` runs for Alice, `self` *is* Alice's object, so `self.name = name` stores `"Alice"` on Alice.
- When `__init__` runs for Bob, `self` *is* Bob's object, so the very same line stores `"Bob"` on Bob.

The same code runs for every student, but `self` points to a different object each time. That is how one blueprint produces many different objects.

````python
print(s1)
````

## 3. Class vs. object: what is the difference?

| | **Class** | **Object** (instance) |
|---|---|---|
| What is it? | A blueprint / template / definition | A concrete thing built from the blueprint |
| How many? | Written **once** | Create **as many as you want** |
| Created by | `class Student:` ... | `Student("Alice", "Finance", 3.8)` |
| Contains | The *description* of the attributes and the *code* for the methods | The *actual values* of the attributes (`"Alice"`, `"Finance"`, `3.8`) |
| Analogy | Blank loan application form | John's completed loan application |
| In our code | `Student` | `s1`, `s2`, `s3` |

**One sentence to remember:** *the class is the recipe; the objects are the cakes.*

Python can tell us which class an object came from — and this reveals that `str`, `list`, and `float` are classes too:

````python
print(type(s1))                      # which class did s1 come from?
print(isinstance(s2, Student))       # "is s2 a Student?" -> True
print(s1 is s2)                      # same object? -> False: two different students

print(type("aapl"))                  # the string "aapl" is an object of the class str
print(type([1, 2, 3]))               # this list is an object of the class list
````

**In Python, everything is an object**, and every object comes from some class. The only new thing today is that we are writing the classes ourselves.

## 4. What are methods, and how do we call them?

> **A method is a function that lives inside a class and operates on an object's data.**

Compared with an ordinary function, a method:

1. is defined **inside** the class (indented under `class ...:`);
2. always has `self` as its **first parameter**;
3. is called **on an object**, using dot notation: `s1.study()`.

````python
s1.study()      # call the study() method ON the object s1
s2.study()      # same method, different object -> different output
s1.play()
````

Notice: `study` is defined as `def study(self):` — **one parameter** — but we call it with **zero arguments**: `s1.study()`. Python fills in `self` for you. Behind the scenes:

```text
    s1.study()        is the same as        Student.study(s1)
                                                         ^^
                                                    this becomes "self"
```

You can even write the long form yourself (nobody does in practice, but it shows what is going on):

````python
Student.study(s1)       # exactly the same as s1.study()
````

### Attributes vs. methods: the parentheses matter

- `s1.gpa` is an **attribute**: a piece of data. **No parentheses.**
- `s1.study()` is a **method**: an action. **Parentheses required**, even with no arguments.

Forget the parentheses and Python does not complain — it just quietly does nothing useful:

````python
print(s1.study)       # forgot the () -> Python shows you the method itself; nothing runs
s1.study()            # with the ()   -> the method actually runs
````

Methods can also take **extra parameters** after `self` and can **return values** instead of printing. The next example shows both.

## 5. Example: the `Student` class

A fuller `Student` class. New things here:

- A **default value** in `__init__` (`gpa=0.0`), so a student can be created without a GPA.
- An attribute that is **not** passed in but set automatically (`hours_studied` starts at 0).
- Methods with **parameters** (`study(hours)`, `change_major(new_major)`), a method that **returns** a value (`is_honor_student()`), and `__str__`.

````python
class Student:
    """A college student."""

    def __init__(self, name, major, gpa=0.0):     # gpa has a DEFAULT value
        self.name = name
        self.major = major
        self.gpa = gpa
        self.hours_studied = 0                    # not passed in: every new student starts at 0

    def study(self, hours):
        """MODIFIES the object: each hour raises GPA by 0.05 (max 4.0)."""
        self.hours_studied += hours
        self.gpa = min(4.0, self.gpa + 0.05 * hours)
        print(f"{self.name} studied {hours} hour(s). GPA: {self.gpa:.2f}")

    def change_major(self, new_major):
        """MODIFIES the object."""
        print(f"{self.name} is switching from {self.major} to {new_major}.")
        self.major = new_major

    def is_honor_student(self):
        """REPORTS: returns True/False and changes nothing."""
        return self.gpa >= 3.5

    def __str__(self):
        """Tells print() what to show for this object. Must RETURN a string."""
        return f"Student(name='{self.name}', major='{self.major}', gpa={self.gpa:.2f})"
````

### Two kinds of methods

| Kind | What it does | Here |
|---|---|---|
| **Modifies** the object | Changes one or more attributes (and may also print a message) | `study()`, `change_major()` |
| **Reports** on the object | Returns a value computed from the attributes, changing nothing | `is_honor_student()`, `__str__()` |

### Creating the objects

````python
alice = Student("Alice", "Finance", 3.4)
bob   = Student("Bob", "Accounting", gpa=3.1)
chen  = Student("Chen", "Economics")            # gpa uses the default, 0.0

print(alice)      # thanks to __str__, print() now shows something readable
print(bob)
print(chen)
````

> **About `__str__`:** like `__init__`, it is a special "double-underscore" method that Python calls for you — whenever you `print()` the object. You never call `alice.__str__()` yourself.

### Calling methods on the objects

````python
alice.study(3)            # we pass ONE argument (hours); Python passes self automatically
alice.study(4)
print("Honor student?", alice.is_honor_student())
print("Hours studied:", alice.hours_studied)

print()
bob.change_major("Finance")   # Bob is a separate object; Alice's studying did nothing to him
print(bob)
````

### Where OOP really shines: many objects, one loop

Because every student has the same methods, we can put students in a list and treat them uniformly.

````python
students = [alice, bob, chen, Student("Dana", "Finance", 3.55)]

for s in students:            # everybody studies for 2 hours
    s.study(2)

print("\nHonor roll:")
for s in students:
    if s.is_honor_student():
        print("  *", s.name, f"({s.gpa:.2f})")

print("\nAverage GPA:", round(sum(s.gpa for s in students) / len(students), 2))
````

## 6. Common mistakes (and how to read the error messages)

The first two cells **crash on purpose**. You *will* see these errors in your own code; knowing what they mean saves a lot of time.

### Mistake 1: forgetting `self` in a method definition

````python
class Account:
    def __init__(self, owner):
        self.owner = owner

    def show():                       # <-- WRONG: missing self
        print("An account")

a = Account("Alice")
a.show()      # Python passes a as the first argument, but show() accepts 0 arguments -> TypeError
````

**Fix:** `def show(self):`. Whenever you see `TypeError: ... takes 0 positional arguments but 1 was given`, a missing `self` is the first thing to check.

### Mistake 2: forgetting `self.` when storing an attribute

````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        balance = balance             # <-- WRONG: just a local variable; it vanishes when __init__ ends

a = Account("Alice", 100)
print(a.balance)      # AttributeError: the object never got a balance attribute
````

**Fix:** `self.balance = balance`. Data is only remembered on the object if you store it through `self`.

### Mistake 3: forgetting the parentheses (no crash, just silent confusion)

````python
print(alice.is_honor_student)       # the method itself, not True/False
print(alice.is_honor_student())     # the answer we wanted
````

If a method "does nothing" or you see `<bound method ...>` where you expected a value, check for missing parentheses.

## 7. Summary

| Term | Meaning | In code |
|---|---|---|
| **Class** | A blueprint describing a kind of thing: its attributes and methods | `class Student:` |
| **Object** / **instance** | A concrete thing created from a class | `s1 = Student("Alice", "Finance", 3.8)` |
| **Instantiation** | Creating an object from a class | calling `Student(...)` |
| **Attribute** | A piece of data stored on an object | `s1.gpa` |
| **Method** | A function defined inside a class that acts on an object | `s1.study(2)` |
| `__init__` | The initializer; runs **automatically, once, when the object is created**; sets up the attributes | `def __init__(self, ...):` |
| `self` | "This particular object"; the first parameter of every method; Python fills it in for you | `self.name = name` |
| **Dot notation** | How you reach attributes and methods | `object.attribute`, `object.method()` |
| `__str__` | Tells Python what `print(object)` should display; must return a string | `def __str__(self): return "..."` |

**Key takeaways**

1. A **class** bundles data (attributes) and behavior (methods) into one unit.
2. **Objects** are created by calling the class like a function: `Student(...)`. Each object has its own attribute values.
3. `__init__` is called **automatically** every time an object is created, and never when the class is merely defined.
4. **Methods** are called on objects with dot notation. Python passes the object as `self`; you supply only the other arguments.
5. Don't forget: `self` in every method definition, `self.` when storing attributes, and `()` when calling methods.

### Practice Example 1: `BankAccount`

**What the class should be.** A `BankAccount` represents one savings account. Every account knows who owns it, how much money is in it, and what interest rate it earns. An account can accept deposits and withdrawals — but it must refuse a withdrawal larger than the balance.

**Attributes**

| Attribute | Meaning | Example value |
|---|---|---|
| `owner` | name of the account holder | `"Alice"` |
| `balance` | money currently in the account | `1000` |
| `annual_rate` | yearly interest rate, as a decimal | `0.04` (= 4%) |

**Methods** (plus `__init__`)

| Method | Kind | What it does |
|---|---|---|
| `deposit(amount)` | modifies | adds `amount` to the balance and prints the new balance |
| `withdraw(amount)` | modifies | subtracts `amount` from the balance; refuses (prints a message, changes nothing) if `amount` is larger than the balance |
| `__str__()` | reports | tells `print()` what to show |

**The code**

````python
class BankAccount:
    """One savings account."""

    def __init__(self, owner, balance, annual_rate):
        self.owner = owner
        self.balance = balance
        self.annual_rate = annual_rate       # 0.04 means 4% per year

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"{self.owner} deposited ${amount:.2f}. Balance: ${self.balance:.2f}")

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"{self.owner} cannot withdraw ${amount:.2f}: balance is only ${self.balance:.2f}.")
            return                           # leave the method early; nothing changes
        self.balance = self.balance - amount
        print(f"{self.owner} withdrew ${amount:.2f}. Balance: ${self.balance:.2f}")

    def __str__(self):
        return f"BankAccount(owner={self.owner}, balance={self.balance:.2f}, rate={self.annual_rate})"
````

### Practice Example 2: `Stock`

**What the class should be.** A `Stock` represents shares of one company that you own. It knows the ticker symbol, how many shares you have, the price you paid per share, and the current price. The current price can be updated as the market moves, and the stock can report what your shares are worth right now.

**Attributes**

| Attribute | Meaning | Example value |
|---|---|---|
| `ticker` | the stock symbol | `"AAPL"` |
| `shares` | number of shares you own | `50` |
| `buy_price` | price per share you paid | `190.00` |
| `current_price` | latest price per share — **not passed in**; starts equal to `buy_price` | `190.00` |

**Methods** (plus `__init__`)

| Method | Kind | What it does |
|---|---|---|
| `update_price(new_price)` | modifies | stores the latest price in `current_price` |
| `market_value()` | reports | returns `shares × current_price` |
| `__str__()` | reports | tells `print()` what to show |

**The code**

````python
class Stock:
    """Shares of one company that you own."""

    def __init__(self, ticker, shares, buy_price):
        self.ticker = ticker
        self.shares = shares
        self.buy_price = buy_price
        self.current_price = buy_price       # not passed in: starts equal to buy_price

    def update_price(self, new_price):
        self.current_price = new_price

    def market_value(self):
        return self.shares * self.current_price

    def __str__(self):
        return f"{self.ticker}: {self.shares} shares, bought at {self.buy_price:.2f}, now {self.current_price:.2f}"
````

### Solution 1.3

````python
class BankAccount:
    """One savings account."""

    def __init__(self, owner, balance, annual_rate):
        self.owner = owner
        self.balance = balance
        self.annual_rate = annual_rate

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"{self.owner} deposited ${amount:.2f}. Balance: ${self.balance:.2f}")

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"{self.owner} cannot withdraw ${amount:.2f}: balance is only ${self.balance:.2f}.")
            return
        self.balance = self.balance - amount
        print(f"{self.owner} withdrew ${amount:.2f}. Balance: ${self.balance:.2f}")

    def add_interest(self):                                  # NEW
        interest = self.balance * self.annual_rate           # one year of interest
        self.balance = self.balance + interest               # add it to the balance
        print(f"Interest ${interest:.2f} added. Balance: ${self.balance:.2f}")

    def __str__(self):
        return f"BankAccount(owner={self.owner}, balance={self.balance:.2f}, rate={self.annual_rate})"


acct = BankAccount("Chen", 1000, 0.05)
acct.add_interest()       # $50.00  -> 1050.00
acct.add_interest()       # $52.50  -> 1102.50
print(acct.balance)
````

**Explanation.** `add_interest` is a *modifying* method with no extra parameters: everything it needs (`balance`, `annual_rate`) is already stored on the object, reachable through `self`. The second year's interest (52.50) is larger than the first (50.00) because the method uses the *current* `self.balance`, which already includes the first year's interest. Compound interest comes for free from storing state on the object.

````python
class Stock:
    """Shares of one company that you own."""

    def __init__(self, ticker, shares, buy_price):
        self.ticker = ticker
        self.shares = shares
        self.buy_price = buy_price
        self.current_price = buy_price

    def update_price(self, new_price):
        self.current_price = new_price

    def market_value(self):
        return self.shares * self.current_price

    def profit_loss(self):                                   # NEW
        return self.shares * (self.current_price - self.buy_price)

    def __str__(self):
        return f"{self.ticker}: {self.shares} shares, bought at {self.buy_price:.2f}, now {self.current_price:.2f}"


# re-create the objects so they come from the new blueprint
aapl = Stock("AAPL", 50, 190.00)
msft = Stock("MSFT", 20, 410.00)
nvda = Stock("NVDA", 15, 118.00)
aapl.update_price(205.50)
msft.update_price(395.25)
nvda.update_price(135.00)

portfolio = [aapl, msft, nvda]

for stock in portfolio:
    print(stock.ticker, "profit/loss:", stock.profit_loss())

print("Winners:")
for stock in portfolio:
    if stock.profit_loss() > 0:
        print("  ", stock.ticker)
````

**Explanation.** `profit_loss` is a *reporting* method: it returns a number and changes nothing, so its result can be printed or compared (`> 0`) directly. AAPL gains 775.0, MSFT loses 295.0, NVDA gains 255.0, so the winners are AAPL and NVDA. The objects are re-created after editing the class because the old ones were built from the old blueprint and have no `profit_loss` method.
