## The `StockAccount` class

**Attributes** — every account object has these six pieces of data:

| Attribute | Meaning | Where it changes |
|---|---|---|
| `owner` | name of the account holder | set in `__init__`, never changes |
| `cash` | cash in the account, in dollars | set in `__init__`; changed by `deposit`, `buy`, `sell` |
| `prices` | the market: a dictionary `ticker → price per share` | set in `__init__`, never changes |
| `holdings` | a dictionary `ticker → number of shares owned`; starts at 0 for all three stocks | changed by `buy`, `sell` |
| `stocks_value` | total market value of all shares owned, in dollars; starts at 0 | changed by `buy`, `sell` |
| `total_value` | `cash + stocks_value`: everything the account is worth; starts equal to the opening cash | changed by `deposit` only — a trade moves money between cash and stocks, so it does not change the total |

**Methods** — what an account can do:

| Method | Kind | What it does | Prints? |

| `sell(ticker, shares)` | modifies | sells `shares` of `ticker` at the market price: cash goes up, holdings and stocks value go down | never |
| `__str__()` | reports | returns a one-line summary of the account, used by `print(account)` | never — it *returns* text |
