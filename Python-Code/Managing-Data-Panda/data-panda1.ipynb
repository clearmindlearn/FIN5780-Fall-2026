# Section 1. Constructing DataFrames

## 1.1 Create a DataFrame from a dictionary of lists (most common)

A **DataFrame** is a **2D table** with **rows + columns** (similar to an Excel sheet).

A common way to build a DataFrame:
- Create a **dictionary**
- Each **key** becomes a **column name**
- Each **value** is a **list** (all lists must have the same length)

**Example:** a small portfolio of six stocks, with the number of shares held and the share price at the end of January 2020.

````python
import pandas as pd  # pd is an alias

# Dictionary of lists:
# - keys   -> column names
# - values -> lists of column values
data = {
    "Ticker": ["AVLO", "CRVN", "IRVB", "ORLN", "TMSN", "ESTF"],
    "Company": ["Arvelo Systems", "Corvane Software", "Ironvale Bank",
                "Orlene Pharma", "Tamsin Foods", "Eastfell Energy"],
    "Shares": [50, 20, 100, 80, 40, 120],
    "Price": [186.66, 285.48, 93.30, 97.20, 191.62, 82.58]
}

df = pd.DataFrame(data)  # DataFrame() converts a dictionary to a DataFrame
df
````

### DataFrame basics: `shape`, `columns`, `index`

````python
# df.shape -> (number of rows, number of columns)
df.shape
````

````python
# df.columns -> column names
df.columns
````

````python
# df.index -> row labels (row labels are called the index)
df.index
````

✅ **Important note about the index**

- In a DataFrame, **row labels are called the index**.
- Even if it looks like `0, 1, 2, ...`, it is still a **label**, not a real "data column".
- In many financial datasets, an ID (like a ticker) is often used as the index.

````python
# set_index("Ticker"):
# - moves Ticker from a normal column to the index (row labels)
# - Ticker will no longer appear as a column unless we reset_index()
df2 = df.set_index("Ticker")
df2
````

### Quick note: `.loc` (label-based lookup)

`.loc[...]` selects rows by **label** (index label).

We will cover `.loc` and `.iloc` more deeply in Section 2 when we discuss **row subsetting/slicing**.

````python
# .loc selects by LABEL (the index label is "IRVB")
df2.loc["IRVB"]
````

````python
# reset_index() moves the index back to a normal column
# It returns a NEW DataFrame, so we save the result in df3
df3 = df2.reset_index()
df3
````

## 1.2 Series vs DataFrame (and what about the index?)

### Big picture
- **Series** = **1D** data (like one column) + **index (row labels)**
- **DataFrame** = **2D** table + **index (row labels)** + **columns (column labels)**

### 1.2.1 Create a Series (simple example)

````python
# A Series is 1D: values + index (row labels)
prices = pd.Series(
    [186.66, 285.48, 93.30],
    index=["AVLO", "CRVN", "IRVB"]  # index = row labels
)

prices
````

````python
# A Series also has an index (row labels)
prices.index
````

### 1.2.2 Create a Series from a dictionary

When you create a Series from a dictionary:
- dictionary **keys** become the **index**
- dictionary **values** become the **data**

````python
# keys become the index (row labels)
prices2 = pd.Series({
    "AVLO": 186.66,
    "CRVN": 285.48,
    "IRVB": 93.30
})

prices2
````

### 1.2.3 One column from a DataFrame becomes a Series

**Point:** A DataFrame has multiple columns. If you select one column, you get a Series.

````python
df_small = pd.DataFrame({
    "Company": ["Arvelo Systems", "Corvane Software", "Ironvale Bank"],
    "Price": [186.66, 285.48, 93.30]
}, index=["AVLO", "CRVN", "IRVB"])  # index = row labels

df_small
````

````python
# One bracket -> Series
col_series = df_small["Price"]
print(col_series)

print(type(col_series))
````

````python
# The Series keeps the same row labels (index)
col_series.index
````

````python
# Double brackets -> DataFrame
col_df = df_small[["Price"]]
print(type(col_df))

col_df
````

✅ **Index reminder**
- For a **DataFrame**, **row labels are the index** (`df.index`)
- For a **Series**, labels are also the **index** (`series.index`)

## 2.2 Row subsetting / slicing (loc, iloc)

### 2.2.1 `.iloc` = integer position-based

`.iloc` uses **positions**: 0th row, 1st row, 2nd row, ...

````python
# First row by position
df.iloc[0]
````

````python
# First 5 rows by position (0,1,2,3,4)
df.iloc[0:5]
````

````python
# Specific rows by position
df.iloc[[0, 3, 10]]  # multiple rows --> use double brackets!!
````

````python
# Rows 30 to 34 by position: the first five companies in February 2020
df.iloc[30:35]
````

### 2.2.2 `.loc` = label-based (IMPORTANT: 0,1,2 are labels, not "numbers")

`.loc` selects by **label** (index label).

When your DataFrame has the default index, the labels look like `0, 1, 2, ...`, but **they are still labels**, not "row numbers".

````python
# Check the row labels (index)
df.index
````

````python
# .loc uses LABELS (index labels)
df.loc[0]
````

````python
# .iloc uses POSITIONS (0th row, 1st row, ...)
df.iloc[0]
````

✅ With the default index, **label 0 happens to match position 0**, so `.loc[0]` and `.iloc[0]` return the same row.

But they are conceptually different.

### Slicing difference (common confusion)
- `.iloc[a:b]` excludes `b`
- `.loc[a:b]` includes `b` (because these are labels)

````python
# iloc: stop is EXCLUDED
df.iloc[0:4]   # positions 0,1,2,3
````

````python
# loc: stop is INCLUDED (because these are labels)
df.loc[0:4]    # labels 0,1,2,3,4
````

### 2.2.3 Show the REAL difference: set `RecordID` as the index

Now `.loc[...]` uses **RecordID labels**, while `.iloc[...]` still uses **positions**.

✅ **Why `RecordID` and not `Ticker`?** Each company appears in three months, so a ticker does not point to one row. `RecordID` (ticker-year-month) is different in every row.

````python
# Make RecordID the index (row labels)
df_id = df.set_index("RecordID")
df_id.head()
````

````python
df_id.shape  # 16 columns (df) ---> 15 columns (df_id) since RecordID becomes the index
````

````python
# Now the labels are record IDs (not 0,1,2,...)
df_id.index
````

````python
# loc: label-based (RecordID label)
df_id.loc["ALDG-2020-01"]
````

````python
# iloc: position-based (first row in the table)
df_id.iloc[0]
````

````python
# Several labels at once --> use a list (double brackets)
# Arvelo Systems in January, February and March 2020
df_id.loc[["AVLO-2020-01", "AVLO-2020-02", "AVLO-2020-03"]]
````

````python
# This FAILS because label 0 does not exist anymore:
df_id.loc[0]   # Error!
````

✅ **Summary**
- `.loc` → **label-based**
- `.iloc` → **position-based**
