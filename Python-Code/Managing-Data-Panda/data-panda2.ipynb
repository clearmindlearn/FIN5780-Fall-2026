# Section 4. Row subsetting using boolean filters

## What is Row Filtering Using Boolean Expressions?

**Row filtering** (also called **boolean indexing**) is a way to select only the rows from a DataFrame that satisfy a given condition.

When you write a comparison like `df["Price"] >= 100`, Pandas evaluates it for **every row** and returns a **Series of True/False values** (one per row). When you place that Series inside `df[...]`, Pandas keeps only the rows where the value is `True` and drops the rows where it is `False`.

**Typical syntax:**

```
df[ boolean_expression ]
```

How it works step-by-step:

| Step | Code | Result |
|------|------|--------|
| 1. Write a condition | `df["Price"] >= 100` | A Series of `True` / `False` for every row |
| 2. Pass it into `df[...]` | `df[ df["Price"] >= 100 ]` | A new DataFrame containing **only** the rows where the condition is `True` |

**Think of it as:** `df[ True/False for every row ]` → returns only the `True` rows.

````python
df.head()
````

````python
boolean_exp = df["Price"] >= 100  # --> boolean expression/filter
boolean_exp
````

````python
# df[ boolean_exp ]
df[df["Price"] >= 100]
````

````python
# how many rows have a Price of at least 100?
df[df["Price"] >= 100].shape
````

````python
len(df[df["Price"] >= 100])
````

# Section 5. Creating a new column

## 5.1 A new column from one existing column

A new column can be **calculated from existing columns**.

**Typical syntax:**

```
df["NewColumn"] = calculation using existing columns
```

- The **left side** is the name of the new column.
- The **right side** is the calculation. Pandas does the calculation for **every row**.

**Example:** `MarketCap` is in millions of dollars. Create a new column that shows it in **billions** of dollars (divide by 1,000).

````python
df["MarketCapBillions"] = df["MarketCap"] / 1000  # adds a new column to df
df[["Company", "MarketCap", "MarketCapBillions"]].head()
````

## 5.2 A new column from two existing columns

The calculation can use **two (or more)** columns.

**Example:** profit margin (percent) = `NetIncome` ÷ `Revenue` × 100

````python
df["ProfitMargin"] = df["NetIncome"] / df["Revenue"] * 100
df[["Company", "Revenue", "NetIncome", "ProfitMargin"]].head()
````

````python
# df now has 18 columns: the 16 original columns + 2 new columns
df.shape
````

✅ **Important**
- Creating a new column **changes `df` itself**. (A filter does not change `df`, but a new column does.)
- The new column is added at the **right end** of the table.
- If a value is missing (`NaN`) in the calculation, the new value is also missing.

## 5.3 Use the new column like any other column

A new column works with everything from Sections 3 and 4: statistics and filters.

````python
df["ProfitMargin"].mean()  # the average profit margin (percent)
````

````python
# Filter on the new column: company-months with a negative profit margin
df[df["ProfitMargin"] < 0]
````

## ✅ Summary

- `df["col"]` → one column as a **Series**; `df[["col"]]` → one column as a **DataFrame**
- `df[["col1", "col2"]]` → several columns (double brackets)
- `df["col"].min()`, `.max()`, `.mean()`, `.sum()` → basic statistics
- `df[ condition ]` → only the rows where the condition is `True`
- `df["NewColumn"] = calculation` → adds a **new column** to `df`
